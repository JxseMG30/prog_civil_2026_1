# *Dosificación de Mezcla para Columnas*

*Planteamiento:*

Un ingeniero residente necesita calcular cuántos sacos de cemento se requieren para vaciar una columna de sección circular.

Se sabe que para un concreto estándar de 210 kg/cm2 (3000 psi), se consumen 9.5 sacos de cemento por cada metro cúbico de mezcla.

*Requerimientos:*

1. Solicitar la altura de la columna en metros.
2. Solicitar el diámetro de la columna en metros.
3. Calcular el volumen de la columna: $$v= \pi * r^2 * h$$
. (Sugerencia: usar 3.1416
 o importar math.pi).
4. Calcular la cantidad de sacos:
$$Sacos=v * 9.5$$

*Prueba del programa:*

Para una columna de 3 metros de alto y 0.4 metros de diámetro (r=0.2
):

Volumen ≈0.377 m3
.
Sacos necesarios ≈3.58
 (el estudiante verá aquí la importancia de los tipos decimales).

````python
#[ENTRADA]: Solicitar la altura de la columna en metros
 # Solicitar el diámetro de la columna en metros.
h = float(input("Ingrese la altura de la columna en metros: "))
d = float(input("Ingrese el diámetro de la columna en metros: "))
# PROCESO: valcula el radio, calcula el volumen de la columna, calcula sacos
r = d/2
v = 3.14 * r**2 * h
sacos = v * 9.5
 # SALIDA: imprime el resultado
print("Volumen ≈", v, "m3")
print("Sacos necesarios ≈", sacos, "m3")
#
````

```markdown
El código para calcular los sacos de cemento realiza las siguientes operaciones:

*   `h = float(input("Ingrese la altura de la columna en metros: "))`
    *   Esta línea solicita al usuario que ingrese la altura de la columna en metros. La función `input()` lee la entrada del usuario como una cadena de texto, y `float()` la convierte a un número decimal (flotante) para poder realizar cálculos matemáticos. El valor se almacena en la variable `h`.

*   `d = float(input("Ingrese el diámetro de la columna en metros: "))`
    *   De manera similar a la línea anterior, esta línea solicita al usuario el diámetro de la columna en metros. La entrada se convierte a un número flotante y se guarda en la variable `d`.

*   `r = d/2`
    *   Aquí se calcula el radio de la columna. El radio `r` es la mitad del diámetro `d`.

*   `v = 3.14 * r**2 * h`
    *   Esta línea calcula el volumen de la columna circular utilizando la fórmula `V = π * r^2 * h`. Se utiliza `3.14` como una aproximación de Pi, `r**2` calcula el radio al cuadrado, y el resultado se multiplica por la altura `h`. El volumen calculado se guarda en la variable `v`.

*   `sacos = v * 9.5`
    *   Se calcula la cantidad de sacos de cemento necesarios. Esto se hace multiplicando el volumen de la columna (`v`) por la cantidad de sacos requeridos por metro cúbico de mezcla, que es `9.5` (según el planteamiento del problema). El resultado se almacena en la variable `sacos`.

*   `print("Volumen ≈", v, "m3")`
    *   Esta línea imprime en la consola el volumen calculado de la columna. Muestra el texto "Volumen ≈", seguido del valor de la variable `v` y la unidad "m3".

*   `print("Sacos necesarios ≈", sacos, "m3")`
    *   Finalmente, esta línea imprime la cantidad de sacos de cemento necesarios. Muestra el texto "Sacos necesarios ≈", el valor de la variable `sacos` y la unidad "m3" (aunque los sacos no se miden en m3, se entiende como la cantidad de sacos por el volumen).
```
