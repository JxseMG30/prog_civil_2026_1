## **Teoría del Confort Térmico en Ganado Vacuno**

El confort térmico en el ganado vacuno se refiere a un estado de equilibrio en el que los animales no experimentan estrés por calor o frío, permitiéndoles mantener su temperatura corporal central dentro de un rango óptimo. Este equilibrio es crucial para la salud, el bienestar y la productividad de los animales, especialmente en sistemas de producción de leche y carne.

### **Zona de Termoneutralidad (ZTN)**

Los bovinos, al igual que otros animales de sangre caliente, poseen una Zona de Termoneutralidad (ZTN), que es el rango de temperaturas ambientales dentro del cual pueden mantener su temperatura corporal constante sin necesidad de aumentar su tasa metabólica para producir calor o disiparlo. Dentro de esta zona, el animal no gasta energía extra para regular su temperatura.

### **Estrés por Calor**

Cuando la temperatura ambiental excede la ZTN, el ganado experimenta estrés por calor. Este fenómeno es una de las principales preocupaciones en la ganadería, especialmente en regiones tropicales y subtropicales. Los síntomas del estrés por calor incluyen:

*   **Aumento de la frecuencia respiratoria (jadeo):** Principal mecanismo para disipar calor evaporativo.
*   **Disminución del consumo de alimento:** Para reducir la producción interna de calor.
*   **Reducción de la producción de leche y carne:** Por la redirección de energía a la termorregulación y la disminución del consumo.
*   **Alteraciones reproductivas:** Afecta la fertilidad y la tasa de concepción.
*   **Comportamientos alterados:** Búsqueda de sombra, aumento del consumo de agua, reducción de la actividad.

### **Estrés por Frío**

Aunque menos común en el contexto de Sucre, el estrés por frío ocurre cuando la temperatura ambiental cae por debajo de la ZTN. Los animales responden aumentando su metabolismo para generar calor, lo que también impacta negativamente la productividad y la salud. Los síntomas incluyen:

*   **Escalofríos y piloerección:** Mecanismos para generar y retener calor.
*   **Aumento del consumo de alimento:** Para obtener más energía para la termogénesis.
*   **Disminución de la producción.**

### **Factores que Influyen en el Confort Térmico**

Además de la temperatura del aire, otros factores meteorológicos influyen significativamente en el confort térmico del ganado:

*   **Humedad Relativa:** Una alta humedad reduce la eficacia del enfriamiento evaporativo (sudoración y jadeo), intensificando el estrés por calor.
*   **Radiación Solar:** La exposición directa al sol aumenta la carga térmica del animal.
*   **Velocidad del Viento:** El viento puede ayudar a la disipación de calor por convección, pero un viento frío puede acentuar el estrés por frío.
*   **Precipitación:** La lluvia, especialmente si es fría, puede contribuir al estrés por frío.

### **Índices de Confort Térmico**

Para cuantificar el riesgo de estrés térmico, se utilizan índices de confort que combinan diferentes variables meteorológicas en un solo valor. El Índice de Temperatura y Humedad (ITH) es uno de los más utilizados, ya que integra la temperatura de bulbo seco y la humedad, proporcionando una estimación efectiva de la percepción térmica del animal. Estos índices permiten a los productores tomar decisiones informadas para mitigar el impacto del estrés térmico en sus rebaños.

# **Métodos y cálculos**

* Índice de temperatura y humedad (producción intensiva)

$$ITH = 0.72(T_{bs} + T_{bh}) + 40.6$$

Donde:


$T_{bs}$ Temperatura bulbo seco ($^oC$)


$T_{bh}$ Temperatura de bulbo húmedo ($^oC$)

Para la correlación de los datos de la especie bovina en la producción de leche se recomienda
utilizar los límites propuestos por National Weather Service:

Sí $ITH ≤ 70$ ambiente no estresante

Sí $ITH > 70 \space And \space ITH ≤ 78$ situación de alerta por estrés térmico

Sí $ITH > 78 \space And \space ITH ≤ 83$ peligro por estrés térmico

Sí $ITH > 83$ situación crítica

````python
# ENTRADA: Pedir tbs (Temperatura de bulbo secoo) y tbh (Temperatura de bulbo húmedo)
tbs = float(input("Ingrese la temperatura de bulbo seco, Grados Centigrados: "))
tbh = float(input("Ingrese la temperatura de bulbo húmedo, Grados Centigrados: "))

# PROCESO: Calculo de ith (índice de Temperatura y Humedad)
ith = 0.72 * (tbs + tbh) + 40.6

# SALIDA: Imprimir resultado de la evaluación del ith y el confort térmico
if ith <= 70:
  print("El índice de temperatura y humedad es: ", ith)
  print("Ambiente no estresante")

elif ith > 70 and ith <= 78:
  print("El índice de temperatura y humedad es: ", ith)
  print("Situación de alerta por estrés térmico")

elif ith > 78 and ith <= 83:
  print("El índice de temperatura y humedad es: ", ith)
  print("peligro por estrés térmico")

elif ith > 83:
  print("El índice de temperatura y humedad es: ", ith)
  print("Situación crítica")
````
